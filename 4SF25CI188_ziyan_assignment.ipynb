```python
index = pd.MultiIndex.from_tuples(
    [('A', 'X'), ('A', 'Y'),
     ('B', 'X'), ('B', 'Y'),
     ('C', 'X'), ('C', 'Y')],
    names=['category', 'subcategory']
)
```

```python
df=pd.DataFrame({
    'value':np.random.randint(10,100,size=6)

},index=index)
print(df)
```

```python
result = df.groupby(level=['category', 'subcategory'])['value'].sum()

print(result)
```
